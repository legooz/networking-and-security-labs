```python
def letterToPosition(letter):
    if letter == ' ':
        return None
    return ord(letter.lower()) - ord('a')

def positionToLetter(position):
    return chr((position % 26) + ord('a'))

def encryptLetter(letter, key):
    if(letter != ' '):
        pos = letterToPosition(letter)
        newLetter = positionToLetter(pos + key)
    else:
        newLetter = ' '
    return newLetter

def decrpytLetter(letter, key):
    if(letter != ' '):
        pos = letterToPosition(letter)
        newLetter = positionToLetter(pos - key)
    else:
        newLetter = ' '
    return newLetter


def encryptMessage(message, key):
    keyIdx = 0
    newMessage = ''

    for i, letter in enumerate(message):
        if letter.isalpha():
            keyNum = letterToPosition(key[keyIdx % len(key)])
            newMessage += encryptLetter(letter, keyNum)
            keyIdx += 1
        else:
            newMessage += letter
    return newMessage

def decryptMessage(message, key):
    keyIdx = 0
    newMessage = ''
    
    for i, letter in enumerate(message):
        if letter.isalpha():
            keyNum = letterToPosition(key[keyIdx % len(key)])
            newMessage += decrpytLetter(letter, keyNum)
            keyIdx += 1
        else:
            newMessage += letter
    return newMessage

def wordToNum(word):
    sum = 0
    for letter in word:
        sum += ord(letter)
    return sum

def encryptKey(messageKey, publicKey): #we are using the same key (Symmetric encryption)

    encrypted = [] #list of encrypted characters
    for char in messageKey: #iterate through each char in messageKey
        newValue = (ord(char) ^ publicKey) % 256 #convert the char to ASCII equivalent  on the key
        encrypted.append(chr(newValue)) #Convert the 0 - 255 value back to a char and add it to encrypted[]
    return "".join(encrypted) #join all characters in encrypted[] into a string


def decryptKey(encryptedKey, privateKey): #we are using the same key (Symmetric encryption)
    
    decrypted = [] #list of decrypted characters
    for char in encryptedKey: #iterate through each char in the encryptedKey
        newValue = (ord(char) ^ privateKey) % 256 #convert the char to ASCII equivalent subtract off the key
        decrypted.append(chr(newValue))  #convert 0 - 255 back into char and add it to decrypted[]
    return "".join(decrypted) #join all character in decrypted[] into a string
```

```python
# Synthetic demonstration fixtures; no original classroom messages or keys are included.
messages = ['hello world', 'a short demo', 'learning ciphers']
messageKeys = ['alpha', 'bravo', 'charlie']
symKey = wordToNum('demo-key')
for message, messageKey in zip(messages, messageKeys):
    encryptedKey = encryptKey(messageKey, symKey)
    encryptedMessage = encryptMessage(message, messageKey)
    recoveredKey = decryptKey(encryptedKey, symKey)
    print(encryptedMessage)
    print(decryptMessage(encryptedMessage, recoveredKey))
```
